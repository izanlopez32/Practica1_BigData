Ejercicio 6 — Decidir el ganador de una partida

-----------------------
EXPLICACIÓN DEL RESULTADO 

Este ejercicio es un poco mas elaborado juntando incluso un if anidado en el apartado del else principal, pero tiene la misma esencia y finalidad que los dos
ejercicios anteriores: devolver un Integer en base al resultado de un condicional

Tenemos los siguientes casos con sus respectivas justificaciones

26 y 20 | Printea 20 y por lo tanto se cumple el primer else if al pasarse handA del numero 21
18 y 22 | Printea 18 y por lo tanto se cumple el segundo else if al pasarse handB del numero 21
24 y 25 | Printea 0 y por lo tanto se cumple el if principal al pasarse ambos de 21
17 y 19 | Printea 19 y se cumple el else del if anidado al no llegar ninguno a 21 y handB ser mayor que handA
21 y 20 | Printea 21 y se cumple el if anidado ubicado en el else al superar handA a handB y al no superar ninguno de los dos el numero 21

-----------------------

```scala
def ganador(handA:Int,handB:Int):Int = 
    if(handA > 21 && handB > 21)
        0
    else if(handA > 21 && handB < 21)
        handB
    else if(handA < 21 && handB > 21)
        handA
    else
        if(handA>handB)
            handA
        else
            handB

println(ganador(26,20))
println(ganador(18,22))
println(ganador(24,25))
println(ganador(17,19))
println(ganador(21,20))
```

Ejercicio 9 — Recorrer un Array con while

-----------------------
EXPLICACIÓN DEL RESULTADO 

El contador en el ejercicio debe ser mutable dado a que en cada vuelta del while este sumará un +1 debido a que pertenece al condicionante del propio while, por
lo que controla cuanto dura el bucle, si no llegasemos a incrementar esta variable estariamos ante un bucle infinito y que no tendria ningun final.

-----------------------

```scala
def bust(puntos:Integer):Boolean = 

    if (puntos > 21)
        true
    else
        false

val manos = Array(17, 24, 21, 19, 26)
var i = 0

while(i<manos.length){
    println(manos(i))
    println(bust(manos(i)))
    i += 1
}
```

En mi solución al ejercicio son mutables tanto los contadores para no hacer bucles infinitos como las siguientes variables:

En mi ejercicio son inmutables las siguientes partes:

Las funciones que utilizo son:

-bust - A peticion del ejercicio.

-validarMano - Funcion que solicita el array que analizaremos junto a un while y un if, si no se cumple bust se añade el valor al array de Auxiliar para luego ser exportado con .take(añadidos) para
    tener todos los valores que no son 0 (esto debido a que iniciamos el Array con todos sus valores en 0, añadidos nos ayuda a que, a la hora de exportar, corte exactamente en el ultimo apartado donde

-mejorPuntuacion - Funcion que solicita un array que previamente ha de haber pasado por validarMano, en dicha funcion buscaremos el valor mas alto con un while y un if donde si la posicion x de la mano otorgada es mas 
                    alta que el numero mas grande existente este numero lo sustituye para pasar a pertenecer tambien a la variable mejor. Finalmente devuelve mejor.

Acerca de las estructuras de control las uso practicamente en todas las partes del codigo, tanto en las funciones usandolas de forma anidada con while e if como usandolas a secas solo con if como es el caso de la funcion bust.

La parte que es mas cercana al estilo funcional es probablemente la funcion mejorPuntuacion, tengo una forma de programar muy imperativa y se nota, de hecho esta misma funcion es imperativa(pero menos que el resto), no cuento

La parte mas cercana al estilo imperativo sin duda es validarMano, es lo que mas se acerca a mi forma de programar la cual se basa en usar muchas variables para guardar diferentes partes del proceso del codigo.

-----------------------

```scala
def bust(puntos:Integer):Boolean = 

    if (puntos > 21)
        true
    else
        false

def validarMano(manos:Array[Int]):Array[Int] = {
    var i = 0
    val auxiliar = new Array[Int](4)
    var añadidos = 0
    while(i<manos.length){
        if(!bust(manos(i))){
            auxiliar(añadidos) = manos(i)
            añadidos +=1
        }
        i += 1
    }
    auxiliar.take(añadidos)
}
def mejorPuntuacion(manos:Array[Int]):Int = {
    var mejor = 0
    var i = 0
    while(i<manos.length){
        if (manos(i) > mejor)
            mejor = manos(i)
        i+=1
    }
    mejor
}

val jugadores = List("Alex", "Chen", "Marta", "Sindhu")
val manos = Array(18, 24, 21, 20)
val manosvalidas = validarMano(manos)
println(mejorPuntuacion(manosvalidas))
val manosRonda2 = Array(22, 19, 20, 21)
val manosvalidas2 = validarMano(manosRonda2)
println(mejorPuntuacion(manosvalidas2))
```
